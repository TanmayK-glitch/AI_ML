### Feature Construction

```python
# Feature construction is the process of creating entirely new columns (features) in your tabular dataset by combining or transforming your existing ones
    # When to use:
        # 1. You have domain knowledge
        # 2. Working with linear models
        # 3. Exploratory analysis shows raw features are weak.

    # When not to use: 
        # 1. You already have thousands of features
        # 2. No proper cross-validation pipeline
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
```

```python
data = pd.read_csv('train.csv')
data.sample(5)
```

```python
x = data.drop(columns=['SalePrice', 'Id'])
y = data['SalePrice']
```

```python
imputer = SimpleImputer(strategy='most_frequent')
x_imputed = pd.DataFrame(imputer.fit_transform(x), columns=x.columns)
```

```python
x_encoded = pd.get_dummies(x_imputed, drop_first=True)
```

```python
np.mean(cross_val_score(LinearRegression(), x_encoded, y, scoring='r2', cv=5))
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x_encoded, y, test_size=0.2, random_state=42)
```
