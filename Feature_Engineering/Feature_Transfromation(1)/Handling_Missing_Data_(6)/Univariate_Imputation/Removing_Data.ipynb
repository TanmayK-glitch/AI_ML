```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('data_science_job.csv')
data.head(5)
```

```python
data.isnull().mean()*100
```

```python
cols = [var for var in data.columns if data[var].isnull().mean() < 0.05 and data[var].isnull().mean() > 0]
cols
```

```python
len(data[cols].dropna()) / len(data)
```

```python
new_data = data[cols].dropna()
data.shape, new_data.shape
```

### Removing from Categorical Columns

```python
data['enrolled_university'].value_counts()

temp = pd.concat([
    # Original 
    data['enrolled_university'].value_counts() / len(data),

    # CCA (Complete Case Analysis)
    new_data['enrolled_university'].value_counts() / len(new_data)
], axis=1)

temp.columns = ['Original', 'CCA']
temp

# In the result we noiced that there is no big change on the data
# so it wont cause any issue, if the values would be a bit more affected the CCA is not good option
```

```python
# Applying on different cat column
data['education_level'].value_counts()

temp2 = pd.concat([
    # Original
    data['education_level'].value_counts() / len(data),

    # CCA
    new_data['education_level'].value_counts() / len(new_data)
], axis=1)

temp2.columns = ['Original', 'CCA']
temp2
```
