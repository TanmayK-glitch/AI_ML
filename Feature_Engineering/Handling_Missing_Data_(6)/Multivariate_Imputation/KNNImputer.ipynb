```python
import numpy as np 
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv('train.csv', usecols=['Age', 'Pclass', 'Fare', 'Survived'])
data.head(5)
```

```python
data.isnull().mean()*100
```

```python
x = data.drop(columns=['Survived'])
y = data['Survived']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.2, random_state=42)
X_train
```

```python
knn = KNNImputer(n_neighbors=2, weights='distance')
X_train_knn = knn.fit_transform(X_train)
X_test_knn = knn.transform(X_test)
```

```python
lr = LogisticRegression()

lr.fit(X_train_knn, Y_train)
y_pred = lr.predict(X_train_knn)

y_pred = lr.predict(X_test_knn)   # 712 predictions
acc = accuracy_score(Y_test, y_pred)  # Y_test has 179 values
acc
```

### Comparing the results with SimpleImputer

```python
si = SimpleImputer(strategy='mean')

X_train_si = si.fit_transform(X_train)
X_test_si = si.transform(X_test)
```

```python
lr = LogisticRegression()

lr.fit(X_train_si, Y_train)
y_pred2 = lr.predict(X_test_si)

acc2 = accuracy_score(y_pred2, Y_test)
acc2
```
