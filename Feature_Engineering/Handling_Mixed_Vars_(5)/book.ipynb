```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('titanic.csv')
data.head(5)
```

```python
data['number'].unique()
```

```python
# wherever you have any numerical and categorical column in same colum, create a new column for Numeric & Categoric data
# Extract Numeric Column
data['number_numeric'] = pd.to_numeric(data['number'], errors='coerce', downcast='integer')
```

```python
# Extract Categorical Column
data['categorical_column'] = np.where(data['number_numeric'].isnull(), data['number'], np.nan)
data.head(5)
```

```python
# Extracting categories for Cabin Column
data['cabin_num'] = data['Cabin'].str.extract('(\d+)') # It captures numerical part
data['cabin_cat'] = data['Cabin'].str[0]
data.head(5)
```

```python
# Extracting  categories for Ticket Column
data['ticket_num'] = data['Ticket'].apply(lambda s : s.split()[-1])
data['ticket_num'] = pd.to_numeric(data['ticket_num'], errors='coerce', downcast='integer')

data['ticket_cat'] = data['Ticket'].apply(lambda s : s.split()[0])
data['ticket_cat'] = np.where(data['ticket_cat'].str.isdigit(), np.nan, data['ticket_cat'])

data.head(5)
```
