```python
import numpy as np
import pandas as pd
```

```python
date = pd.read_csv('orders.csv')
time = pd.read_csv('messages.csv')
```

## Working with Dates

```python
date['date'] = pd.to_datetime(date['date'])
```

### 1. Extract Year

```python
date['date_year'] = date['date'].dt.year
date.sample()
```

### 2. Extract Month

```python
date['date_month'] = date['date'].dt.month
date['date_month_name'] = date['date'].dt.month_name()
date.sample()

```

### 3. Extract Days

```python
date['date_days'] = date['date'].dt.day

# Day of week
date['date_days_week'] = date['date'].dt.day_of_week

# Day of week name
date['day_week_name'] = date['date'].dt.day_name()

# Is weekend??
date['isWeekend'] = np.where(date['day_week_name'].isin(["Sunday", "Saturday"]), 1, 0)
date.sample()
```

### 4. Extract Week of Year

```python
date['date_week'] = date['date'].dt.day_of_week
date.sample()
```

### 5. Extract Quarter

```python
date['quarter'] = date['date'].dt.quarter
date.sample()

# Extract Semester using Quarter
date['semester'] = np.where(date['quarter'].isin([1, 2]), 1, 2)
date.sample()
```

## Working wiht Time

```python
time.info()
time['date'] = pd.to_datetime(time['date'])
```

```python
time['hour'] = time['date'].dt.hour
time['minutes'] = time['date'].dt.minute
time['seconds'] = time['date'].dt.second
```

### 1. Extract Time part

```python
time['time'] = time['date'].dt.time
time.head()
```

### 2. Time difference

```python
import datetime
today = datetime.datetime.today()
today

# In Days
today - time['date']

# In Seconds
(today - time['date'])/np.timedelta64(1, 's')

# In Minutes
(today - time['date'])/np.timedelta64(1, 'm')

# In Hours
(today - time['date'])/np.timedelta64(1, 'h')

# Here timedelta64 tells the difference between the time interval (duration between 2 time)
```
