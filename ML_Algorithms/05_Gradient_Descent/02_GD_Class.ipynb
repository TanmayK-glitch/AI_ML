```python
import numpy as np
from sklearn.datasets import make_regression
```

```python
x,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1, noise=20)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x, y)
print(lr.coef_)
print(lr.intercept_)
```

```python
m = 30.97
```

```python
class GDRegressor:
    def __init__(self, learning_rate, epochs):
        self.m = lr.coef_[0]  # can use hardcoded value too but will affect the y_pred(intercept)
        self.b = -120
        self.learning_rate = learning_rate
        self.epochs = epochs

    def fit(self, x, y):
        for i in range(self.epochs):
            loss = -2 * np.sum(y - self.m * x.ravel() - self.b)
            self.b = self.b - (self.learning_rate * loss)
        print(f"Final intercept (b): {self.b}")
```

```python
gd = GDRegressor(0.001, 100)
```

```python
gd.fit(x, y)
```
