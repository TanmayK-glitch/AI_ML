```python
import numpy as np
from sklearn.datasets import make_regression
```

```python
x,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1, noise=20)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x, y)
print(lr.coef_)
print(lr.intercept_)
```

```python
class GDRegressor:
    def __init__(self, learning_rate, epochs):
        self.m = 100
        self.b = -120
        self.learning_rate = learning_rate
        self.epochs = epochs

    def fit(self, x, y):
        for i in range(self.epochs):
            loss_b = -2 * np.sum(y - self.m * x.ravel() - self.b)
            loss_m = -2 * np.sum((y - self.m * x.ravel() - self.b)* x.ravel())


            self.b = self.b - (self.learning_rate * loss_b)
            self.m = self.m - (self.learning_rate * loss_m)
        print(f"Final intercept (b): {self.b}")
        print(f"Final slope (m): {self.m}")
```

```python
gd = GDRegressor(0.001, 100)
```

```python
gd.fit(x, y)
```
