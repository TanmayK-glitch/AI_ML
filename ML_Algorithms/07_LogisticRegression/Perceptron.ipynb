```python
from sklearn.datasets import make_classification
import numpy as np
import matplotlib.pyplot as plt
```

```python
X, y = make_classification(n_samples=100, n_features=2, n_informative=1, n_redundant=0, 
                           n_classes=2, n_clusters_per_class=1, random_state=42, hypercube=False, class_sep=10)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X, y):
    X = np.insert(X, 0, 1, axis=1)
    weights = np.ones(X.shape[0])
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(1, 100)
        y_hat = step(np.dot(weights, X[j]))
        weights = weights + lr*(y[j] - y_hat) * X[j]

    return weights[0], weights[1:]      
```

```python
def step(z):
    if z > 0:
        return 1
    else:
        return 0
```

## Logistic Regression -- Sigmoid Func

```python
def perceptron_2(X, y):
    np.insert(1, X, y, axis=1)
    weights = np.ones(X.shape[0])
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(1, 100)
        y_hat = sigmoid(np.dot(X[j], weights))
        weights = weights + lr*(y[j] * y_hat) * X[j]

    return weights[0], weights[1:]
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```
